```python
# Calculate the distribution of is_churn
churn_counts = train['is_churn'].value_counts().reset_index()
churn_counts.columns = ['is_churn', 'count']

# Map values for better readability on plot
churn_counts['Label'] = churn_counts['is_churn'].map({0: 'Not Churned (0)', 1: 'Churned (1)'})

# Plotting
plt.figure(figsize=(8, 6))
sns.barplot(x='Label', y='count', data=churn_counts, palette='viridis', hue='Label', legend=False)

# Formatting graph labels
plt.title('Class Imbalance: Churn vs. No Churn', fontsize=14)
plt.xlabel('Status', fontsize=12)
plt.ylabel('Number of Users', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Add exact count values on top of the bars
for index, row in churn_counts.iterrows():
    plt.text(index, row['count'] + (max(churn_counts['count']) * 0.01), f"{row['count']:,}", ha='center', va='bottom', fontsize=11)

plt.tight_layout()
plt.show()
```
